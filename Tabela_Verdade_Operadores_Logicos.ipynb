```python
print("Tabela verdade E lógico")
print("***************************")
resultado = True & True
print("True e True =", resultado)

print("True e False =",True & False)

print("False e True =",False & True)

print("False e False =",False & False)

```

```python
print("\nTabela verdade OU lógico")
print("***************************")
print("True ou True =",True | True)

print("True ou False =",True | False)

print("False ou True =",False | True)

print("False ou False =",False | False)

```

```python
print("\nTabela verdade OU Exclusivo lógico")
print("***************************")
print("True ou exclusivo True =",True ^ True)

print("True ou exclusivo False =",True ^ False)

print("False ou exclusivo True =",False ^ True)

print("False ou exclusivo False =",False ^ False)
```

```python
print("\nTabela verdade Não lógico")
print("***************************")
print("Não True = ", not True)
print("Não False = ", not False)
```
